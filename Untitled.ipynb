```python
# Import the libraries we need
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Loading CSV file
df = pd.read_csv("weatherdataCMSE.csv")

# Keeping only the two columns we are interested in
data = df[['TMIN', 'TMAX']].dropna()

# Choosing the independent and dependent variables
# Here: AWND (wind) predicts PRCP (rain)
X = data[['TMIN']]   # independent variable
y = data['TMAX']     # dependent variable

# Create the model
model = LinearRegression()

# Train (fit) the model
model.fit(X, y)

# Print the results
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("R-squared:", model.score(X, y))

# Make a scatter plot with the regression line
plt.scatter(X, y, alpha=0.3, label="Data Points")
plt.plot(X, model.predict(X), label="Regression Line")
plt.xlabel("Minimum Temperature (TMIN)")
plt.ylabel("Maximum Temperature (TMAX)")
plt.title("Linear Regression: TMIN vs TMAX")
plt.legend()
plt.show()
```

```python
# Import the libraries we need
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Loading CSV file
df = pd.read_csv("weatherdataCMSE.csv")

# Keeping only the two columns we are interested in
data = df[['WSF2', 'AWND']].dropna()

# Choosing the independent and dependent variables
# Here: AWND (wind) predicts PRCP (rain)
X = data[['AWND']]   # independent variable
y = data['WSF2']     # dependent variable

# Create the model
model = LinearRegression()

# Train (fit) the model
model.fit(X, y)

# Print the results
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("R-squared:", model.score(X, y))

# Make a scatter plot with the regression line
plt.scatter(X, y, alpha=0.3, label="Data Points")
plt.plot(X, model.predict(X), label="Regression Line")
plt.xlabel("Average Wind Speed (AWND)")
plt.ylabel("Fastest 2 Minute Wind Speed(WSF2)")
plt.title("Linear Regression: AWND vs WSF2")
plt.legend()
plt.show()
```

```python
import pandas as pd
import statsmodels.api as sm

# Load the data
df = pd.read_csv("weatherdataCMSE.csv")   # your file

# Select variables and drop missing rows
data = df[['TMAX', 'TMIN', 'AWND']].dropna()

# Define X and y
X = data[['TMIN', 'AWND']]   # independent variables
y = data['TMAX']             # dependent variable

# Add constant term for OLS intercept
X = sm.add_constant(X)

# Fit the OLS model
model = sm.OLS(y, X).fit()

# Display the full statistical report
print(model.summary())
```
